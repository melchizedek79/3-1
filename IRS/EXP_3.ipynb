```python
# Input document
document = "The information retrieval system is used to search for information in documents"

# Convert to lowercase
document = document.lower()

# Tokenization
tokens = document.split()

print("Tokens:")
print(tokens)

# List of stop words
stop_words = [
    "a", "an", "the", "is", "are", "am", "and",
    "of", "in", "to", "for", "on", "at", "by"
]

# Stop word removal
filtered_tokens = []

for word in tokens:
    if word not in stop_words:
        filtered_tokens.append(word)

print("\nAfter removing stop words:")
print(filtered_tokens)

# Simple stemming function
def stem(word):

    if word.endswith("ing"):
        return word[:-3]

    elif word.endswith("ed"):
        return word[:-2]

    elif word.endswith("ment"):
        return word[:-4]

    elif word.endswith("tion"):
        return word[:-4]

    elif word.endswith("s") and len(word) > 3:
        return word[:-1]

    return word

# Stemming
stemmed_words = []

for word in filtered_tokens:
    stemmed_words.append(stem(word))

print("\nAfter stemming:")
print(stemmed_words)
```
