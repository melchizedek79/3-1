```python
from sklearn.datasets import fetch_20newsgroups
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
from sklearn.metrics import accuracy_score
```

```python
from sklearn.metrics import classification_report
```

```python
from sklearn.metrics import confusion_matrix
```

```python
categories = {
    'comp.graphics',
    'sci.space',
    'rec.sport.baseball',
    'talk.politics.guns'
}
```

```python
dataset = fetch_20newsgroups(
    subset = 'all',
    categories = categories,
    remove = ('headers', 'footers', 'quotes')
)
```

```python
x = dataset.data
y = dataset.target
```

```python
vectorizer = TfidfVectorizer(stop_words = 'english')
X_features = vectorizer.fit_transform(x)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_features,
    y,
    test_size=0.2,
    random_state=42
)

# Train the classifier
classifier = MultinomialNB()

classifier.fit(X_train, y_train)

# Predict the classes
y_pred = classifier.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy * 100, 2), "%")

# Display the classification report
print("\nClassification Report:\n")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=dataset.target_names
    )
)

# Display the confusion matrix
print("Confusion Matrix:\n")

print(confusion_matrix(y_test, y_pred))

# Test with new documents
new_documents = [
    "NASA is planning a mission to Mars.",
    "The hockey team won the championship.",
    "The new graphics card improves image quality.",
    "The government introduced a new political policy."
]

new_features = vectorizer.transform(new_documents)

predictions = classifier.predict(new_features)

print("\nPredictions:\n")

for document, label in zip(new_documents, predictions):
    print("Document:", document)
    print("Predicted Class:", dataset.target_names[label])
    print()
```
