```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Two sample documents
doc1 = """
Machine learning is a branch of artificial intelligence.
It enables computers to learn from data.
"""

doc2 = """
Artificial intelligence includes machine learning techniques
that allow computers to learn using data.
"""

# Store documents in a list
documents = [doc1, doc2]

# Convert documents into TF-IDF vectors
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

# Compute cosine similarity
similarity_matrix = cosine_similarity(tfidf_matrix)

print("Cosine Similarity Matrix:")
print(np.round(similarity_matrix, 4))

# Display similarity percentage
similarity = similarity_matrix[0][1] * 100
print(f"\nSimilarity between Document 1 and Document 2: {similarity:.2f}%")

# Check if similarity is above 60%
if similarity >= 60:
    print("The documents are more than 60% similar.")
else:
    print("The documents are less than 60% similar.")
```
