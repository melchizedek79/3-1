```python
import xml.etree.ElementTree as ET
import networkx as nx



xml_data = """
<web>
    <page id="A" topic="AI">
        <title>Artificial Intelligence</title>
        <link>B</link>
        <link>C</link>
    </page>

    <page id="B" topic="AI">
        <title>Machine Learning</title>
        <link>C</link>
        <link>D</link>
    </page>

    <page id="C" topic="AI">
        <title>Deep Learning</title>
        <link>A</link>
        <link>D</link>
    </page>

    <page id="D" topic="Sports">
        <title>Sports News</title>
        <link>E</link>
    </page>

    <page id="E" topic="AI">
        <title>Computer Vision</title>
        <link>A</link>
    </page>
</web>
"""


root = ET.fromstring(xml_data)



G = nx.DiGraph()

for page in root.findall("page"):
    page_id = page.get("id")
    title = page.find("title").text
    topic = page.get("topic")

    G.add_node(page_id, title=title, topic=topic)

    for link in page.findall("link"):
        G.add_edge(page_id, link.text)

print("Web Graph:")
print("Nodes:", list(G.nodes()))
print("Edges:", list(G.edges()))


topic = "AI"

topic_nodes = [
    node for node, data in G.nodes(data=True)
    if data.get("topic") == topic
]

print("\nTopic:", topic)
print("Topic-specific pages:", topic_nodes)


topic_graph = G.subgraph(topic_nodes).copy()


pagerank = nx.pagerank(
    topic_graph,
    alpha=0.85
)



print("\nTopic-Specific PageRank:")

for page, score in sorted(
    pagerank.items(),
    key=lambda x: x[1],
    reverse=True
):
    title = G.nodes[page]["title"]
    print(f"{page} - {title}: {score:.4f}")
```
