```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

# Websites that can be selected
WEBSITES = {
    "1": ("BBC", "https://www.bbc.com"),
    "2": ("Reuters", "https://www.reuters.com"),
    "3": ("CNN", "https://www.cnn.com"),
    "4": ("The Hindu", "https://www.thehindu.com")
}

# Get user input
topic = input("Enter the news topic to search: ")

print("\nSelect websites:")
for key, (name, url) in WEBSITES.items():
    print(f"{key}. {name}")

choice = input("\nEnter website numbers separated by comma (example: 1,2): ")

selected = [x.strip() for x in choice.split(",")]

# Search each selected website
for number in selected:

    if number not in WEBSITES:
        print(f"Invalid website option: {number}")
        continue

    name, website = WEBSITES[number]

    print("\n" + "=" * 60)
    print("Searching:", name)
    print("=" * 60)

    try:
        # Download webpage
        response = requests.get(
            website,
            headers={"User-Agent": "Mozilla/5.0"},
            timeout=10
        )

        soup = BeautifulSoup(response.text, "html.parser")

        # Find links containing the topic
        found = 0

        for link in soup.find_all("a", href=True):

            text = link.get_text(" ", strip=True)

            if topic.lower() in text.lower():

                url = urljoin(website, link["href"])

                print("\nNews:", text)
                print("URL :", url)

                found += 1

                if found == 10:
                    break

        if found == 0:
            print("No matching news stories found.")

    except Exception as e:
        print("Error:", e)
```
